```python
cat > headlamp-values.yaml << EOF
ingress:
  enabled: true
  ingressClassName: nginx   
  hosts:
    - host: headlamp.voip.local
      paths:
        - path: /
          type: Prefix
EOF
```

```python
helm install my-headlamp headlamp/headlamp \
	-f headlamp-values.yaml \
	--namespace headlamp \
	--create-namespace
```

```python
  kubectl create token my-headlamp --namespace headlamp
```

```python
cat > headlamp-ingress.yaml <<EOF
apiVersion: networking.k8s.io/v1
kind: Ingress
metadata:
  name: headlamp-ingress
  namespace: headlamp
  annotations:
    nginx.ingress.kubernetes.io/ssl-redirect: "false"
spec:
  ingressClassName: nginx
  rules:
  - host: headlamp.voip.local
    http:
      paths:
      - path: /
        pathType: Prefix
        backend:
          service:
            name: my-headlamp
            port:
              number: 4466
EOF


kubectl apply -f headlamp-ingress.yaml
```

```python
  kubectl create token my-headlamp --namespace headlamp
```
